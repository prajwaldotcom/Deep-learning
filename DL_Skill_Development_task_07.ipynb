```python
import torch
import torch.nn as nn

torch.manual_seed(1)

x = torch.randn(1, 64, 32, 32)
```

```python
conv1 = nn.Conv2d(64, 32, 1)
y = conv1(x)
y = torch.relu(y)
```

```python
conv2 = nn.Conv2d(32, 32, 3, padding=1, groups=4)
y = conv2(y)
y = torch.relu(y)
```

```python
conv3 = nn.Conv2d(32, 64, 1)
y = conv3(y)
```

```python
y = y + x

y = torch.relu(y)

print("Output shape:", y.shape)
```

```python
parameters = sum(p.numel() for p in conv1.parameters())
parameters += sum(p.numel() for p in conv2.parameters())
parameters += sum(p.numel() for p in conv3.parameters())

print("Number of parameters:", parameters)
```
