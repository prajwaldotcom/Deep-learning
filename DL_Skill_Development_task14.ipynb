```python
import torch
import time

device = "cuda" if torch.cuda.is_available() else "cpu"

x = torch.randn(10000000, device=device)

start = time.time()
```

```python
y = x * torch.sigmoid(x)

if device == "cuda":
    torch.cuda.synchronize()

end = time.time()

print("Device:", device)
print("Output shape:", y.shape)
print("Execution time:", end-start, "seconds")
```
