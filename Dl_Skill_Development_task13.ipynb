```python
import torch
import torch.nn as nn
import torch.nn.functional as F

torch.manual_seed(1)
```

```python
x = torch.randn(4, 1, 50)
```

```python
conv1 = nn.Conv1d(1, 16, kernel_size=3, dilation=1)
```

```python
x1 = F.pad(x, (2, 0))
x1 = torch.relu(conv1(x1))
```

```python
conv2 = nn.Conv1d(16, 32, kernel_size=3, dilation=2)

x2 = F.pad(x1, (4, 0))
x2 = torch.relu(conv2(x2))
```

```python
conv3 = nn.Conv1d(32, 32, kernel_size=3, dilation=4)

x3 = F.pad(x2, (8, 0))
x3 = torch.relu(conv3(x3))
```

```python
output_layer = nn.Conv1d(32, 1, 1)

output = output_layer(x3)

print("Input shape:", x.shape)
print("Output shape:", output.shape)
```
