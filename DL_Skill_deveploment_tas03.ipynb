```python
import numpy as np
```

```python
np.random.seed(1)
```

```python
X = np.random.randn(1, 3, 5, 5)
```

```python
W = np.random.randn(2, 3, 3, 3)
b = np.zeros(2)

padding = 1
stride = 1
```

```python
X_pad = np.pad(
    X,
    ((0,0), (0,0), (padding,padding), (padding,padding))
)

N, C, H, W_in = X.shape
F, _, KH, KW = W.shape

H_out = (H + 2*padding - KH)//stride + 1
W_out = (W_in + 2*padding - KW)//stride + 1
```

```python
cols = []

for n in range(N):
    for i in range(0, H_out):
        for j in range(0, W_out):
            patch = X_pad[
                n,
                :,
                i*stride:i*stride+KH,
                j*stride:j*stride+KW
            ]
            cols.append(patch.reshape(-1))

cols = np.array(cols)
```

```python
W_col = W.reshape(F, -1)
```

```python
output = cols @ W_col.T
```

```python
output = output + b
```

```python
output = output.reshape(N, H_out, W_out, F)
output = output.transpose(0, 3, 1, 2)

print("Output shape:", output.shape)
print(output)
```
