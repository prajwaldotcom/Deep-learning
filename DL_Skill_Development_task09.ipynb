```python
import torch
pred = torch.tensor([[0.5, 0.5, 0.4, 0.4]], requires_grad=True)
target = torch.tensor([[0.52, 0.48, 0.4, 0.4]])
```

```python
px = pred[:, 0]
py = pred[:, 1]
pw = pred[:, 2]
ph = pred[:, 3]

tx = target[:, 0]
ty = target[:, 1]
tw = target[:, 2]
th = target[:, 3]

p_x1 = px - pw/2
p_y1 = py - ph/2
p_x2 = px + pw/2
p_y2 = py + ph/2

t_x1 = tx - tw/2
t_y1 = ty - th/2
t_x2 = tx + tw/2
t_y2 = ty + th/2
```

```python
ix1 = torch.maximum(p_x1, t_x1)
iy1 = torch.maximum(p_y1, t_y1)
ix2 = torch.minimum(p_x2, t_x2)
iy2 = torch.minimum(p_y2, t_y2)

iw = torch.clamp(ix2-ix1, min=0)
ih = torch.clamp(iy2-iy1, min=0)

intersection = iw*ih
```

```python
area_p = pw*ph
area_t = tw*th

union = area_p + area_t - intersection

iou = intersection/(union + 1e-7)

# Center distance
center_distance = (px-tx)**2 + (py-ty)**2
```

```python
cx1 = torch.minimum(p_x1, t_x1)
cy1 = torch.minimum(p_y1, t_y1)
cx2 = torch.maximum(p_x2, t_x2)
cy2 = torch.maximum(p_y2, t_y2)

c2 = (cx2-cx1)**2 + (cy2-cy1)**2 + 1e-7
```

```python
ciou = iou - center_distance/c2

loss = 1 - ciou.mean()

loss.backward()

print("CIoU Loss:", loss.item())
print("Gradient:", pred.grad)
```
