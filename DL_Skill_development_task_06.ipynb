```python
import torch

torch.manual_seed(1)
```

```python
X = torch.randn(2, 5, 4)

input_size = 4
hidden_size = 3
```

```python
Wf = torch.randn(input_size + hidden_size, hidden_size*4) * 0.1
bf = torch.zeros(hidden_size*4)
```

```python
Wb = torch.randn(input_size + hidden_size, hidden_size*4) * 0.1
bb = torch.zeros(hidden_size*4)
```

```python
hf = torch.zeros(2, hidden_size)
cf = torch.zeros(2, hidden_size)

forward_output = []

for t in range(5):
    combined = torch.cat([X[:, t, :], hf], dim=1)

    gates = combined @ Wf + bf

    i, f, o, g = torch.chunk(gates, 4, dim=1)

    i = torch.sigmoid(i)
    f = torch.sigmoid(f)
    o = torch.sigmoid(o)
    g = torch.tanh(g)

    cf = f*cf + i*g
    hf = o*torch.tanh(cf)

    forward_output.append(hf)
```

```python
hb = torch.zeros(2, hidden_size)
cb = torch.zeros(2, hidden_size)

backward_output = []

for t in range(4, -1, -1):
    combined = torch.cat([X[:, t, :], hb], dim=1)

    gates = combined @ Wb + bb

    i, f, o, g = torch.chunk(gates, 4, dim=1)

    i = torch.sigmoid(i)
    f = torch.sigmoid(f)
    o = torch.sigmoid(o)
    g = torch.tanh(g)

    cb = f*cb + i*g
    hb = o*torch.tanh(cb)

    backward_output.append(hb)

backward_output.reverse()
```

```python
output = []

for t in range(5):
    output.append(torch.cat([forward_output[t], backward_output[t]], dim=1))

output = torch.stack(output, dim=1)

print("BiLSTM output shape:", output.shape)
```
