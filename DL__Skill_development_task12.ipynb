```python
import torch
import torch.nn as nn
import torch.nn.functional as F

torch.manual_seed(1)
```

```python
conv1 = nn.Conv2d(1, 16, 3, stride=2, padding=1)
conv2 = nn.Conv2d(16, 32, 3, stride=2, padding=1)
```

```python
deconv1 = nn.ConvTranspose2d(32, 16, 4, stride=2, padding=1)
deconv2 = nn.ConvTranspose2d(16, 1, 4, stride=2, padding=1)
```

```python
image = torch.rand(4, 1, 28, 28)
```

```python
x = torch.relu(conv1(image))
x = torch.relu(conv2(x))
```

```python
x = torch.relu(conv2(x))
x = torch.relu(deconv1(x))
reconstructed = torch.sigmoid(deconv2(x))

print("Original shape:", image.shape)
print("Reconstructed shape:", reconstructed.shape)
```

```python
error = (image - reconstructed)**2
error = error.mean(dim=(1,2,3))

print("Reconstruction error:")
print(error)
```

```python
mu1 = image.mean()
mu2 = reconstructed.mean()

variance1 = image.var()
variance2 = reconstructed.var()

covariance = ((image-mu1)*(reconstructed-mu2)).mean()

ssim = (
    (2*mu1*mu2 + 0.01) *
    (2*covariance + 0.03)
) / (
    (mu1**2 + mu2**2 + 0.01) *
    (variance1 + variance2 + 0.03)
)

ssim_loss = 1 - ssim
```
