```python
import torch

torch.manual_seed(1)

X = torch.randn(8, 4)

gamma = torch.ones(4)
beta = torch.zeros(4)
```

```python
mean = X.mean(dim=0)
variance = X.var(dim=0, unbiased=False)

X_norm = (X - mean) / torch.sqrt(variance + 1e-5)

batch_output = gamma * X_norm + beta

print("Batch Normalization output:")
print(batch_output)
```

```python
dout = torch.ones_like(batch_output)

N = X.shape[0]

dgamma = torch.sum(dout * X_norm, dim=0)
dbeta = torch.sum(dout, dim=0)

dX = (gamma / (N * torch.sqrt(variance + 1e-5))) * (
    N*dout
    - torch.sum(dout, dim=0)
    - X_norm*torch.sum(dout*X_norm, dim=0)
)

print("dGamma:", dgamma)
print("dBeta:", dbeta)
print("dX:", dX)
```

```python
mean_layer = X.mean(dim=1, keepdim=True)
variance_layer = X.var(dim=1, keepdim=True, unbiased=False)

X_layer = (X - mean_layer) / torch.sqrt(variance_layer + 1e-5)

layer_output = gamma * X_layer + beta

print("Layer Normalization output:")
print(layer_output)
```
