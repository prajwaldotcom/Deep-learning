```python
import numpy as np
```

```python
X = np.random.randn(5, 4)
y = np.array([0, 1, 2, 1, 0])
```

```python
W1 = np.random.randn(4, 5) * 0.1
b1 = np.zeros((1, 5))

W2 = np.random.randn(5, 3) * 0.1
b2 = np.zeros((1, 3))
```

```python
Z1 = X @ W1 + b1
A1 = np.maximum(0, Z1)

Z2 = A1 @ W2 + b2
```

```python
Z2 = Z2 - np.max(Z2, axis=1, keepdims=True)
exp_Z2 = np.exp(Z2)
P = exp_Z2 / np.sum(exp_Z2, axis=1, keepdims=True)
```

```python
loss = -np.mean(np.log(P[np.arange(len(y)), y] + 1e-8))
print("Loss:", loss)
```

```python
dZ2 = P.copy()
dZ2[np.arange(len(y)), y] -= 1
dZ2 = dZ2 / len(y)
```

```python
dW2 = A1.T @ dZ2
db2 = np.sum(dZ2, axis=0, keepdims=True)
```

```python
dA1 = dZ2 @ W2.T
dZ1 = dA1 * (Z1 > 0)
```

```python
dW1 = X.T @ dZ1
db1 = np.sum(dZ1, axis=0, keepdims=True)
```

```python
lr = 0.01
W1 = W1 - lr * dW1
b1 = b1 - lr * db1
W2 = W2 - lr * dW2
b2 = b2 - lr * db2

print("Weights updated successfully")
```
